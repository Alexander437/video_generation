## Dataset with prompt

```python
df = pd.read_csv('list_attr_celeba.csv')
df.head()
```

```python
old_keys = ['Arched_Eyebrows', 'Attractive',
       'Bags_Under_Eyes', 'Bald', 'Bangs', 'Big_Lips', 'Big_Nose',
       'Black_Hair', 'Blond_Hair', 'Blurry', 'Brown_Hair', 'Bushy_Eyebrows',
       'Chubby', 'Double_Chin', 'Eyeglasses', 'Goatee', 'Gray_Hair',
       'Heavy_Makeup', 'High_Cheekbones', 'Male', 'Mouth_Slightly_Open',
       'Mustache', 'Narrow_Eyes', 'No_Beard', 'Oval_Face', 'Pale_Skin',
       'Pointy_Nose', 'Receding_Hairline', 'Rosy_Cheeks', 'Sideburns',
       'Smiling', 'Straight_Hair', 'Wavy_Hair', 'Wearing_Earrings',
       'Wearing_Hat', 'Wearing_Lipstick', 'Wearing_Necklace',
       'Wearing_Necktie', 'Young']
```

```python
pos_new_keys = ["Изогнутые брови", "Привлекательность",
       "Мешки под глазами", "Лысый", "Челка", "Большие губы", "Большой нос",
       "Черные волосы", "Светлые волосы", "Размытые", "Коричневые волосы", "Густые брови",
       "Пухлость", "Двойной подбородок", "Очки", "Эспаньолка", "Седые волосы",
       "Сильный макияж", "Высокие скулы", "Мужчина", "Рот слегка открыт",
       "Усы", "Узкие глаза", "Без бороды", "Овальное лицо", "Бледная кожа",
       "Острый нос", "Залысины", "Розовые щеки", "Бакенбарды",
       "Улыбчивость", "Прямые волосы", "Волнистые волосы", "В серьгах",
       "В шляпе", "Губная помада", "Ожерелье",
       "В галстуке", "Молодость"]

neg_new_keys = ["Прямые брови", None,
       None, None, None, "Узкие губы", "Маленький нос",
       None, None, "Четкость", None, "Редкие брови",
       "Стройность", None, None, None, None,
       "Без макияжа", None, "Женщина", None,
       None, "Широкие глаза", None, None, "Загар",
       None, None, None, None,
       "Грусть", None, None, None,
       None, None, None,
       None, "Старость"]
```

```python
desc = {old_key: {1: pos_key, -1: neg_key} for old_key, pos_key, neg_key in zip(old_keys, pos_new_keys, neg_new_keys)}

desc
```

```python
new_desc = {
       "image_id": [],
       "prompt": []
}
for _, row in df.iterrows():
       new_desc["image_id"].append(row['image_id'])
       prompt_list = []
       for key, prompt_keys in desc.items():
              res = prompt_keys[row[key]]
              if res is not None:
                     prompt_list.append(res)
       new_desc["prompt"].append(prompt_list)
```

```python
new_df = pd.DataFrame(new_desc, dtype="object")
new_df.head()
```

```python
dataset = Dataset.from_pandas(new_df)
dataset.save_to_disk("celeba_prompts")
```
